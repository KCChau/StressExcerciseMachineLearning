# Step 1: Divide the data into their blocks
This file spilts the original dataset to their appropriate classes:

## AEROBIC
rpm70, rpm75, rpm80, rpm85

## ANAEROBIC
f: sprint1, sprint2, sprint3, sprint4  
S: sprint1, sprint2, sprint3

## REST:
Baseline, REST1, REST 2

## STRESS
f: Opposite, Real, Subtract, TMCT  
S: Opposite, Real, Stroop, Subtract, TMCT

```python
import csv
import os
import datetime
import pandas as pd
import numpy as np

datafiles=['BVP','HR','EDA','TEMP','ACC']
new_data='Extracted'

def removeSampRate(csv):
  csv.drop([0],axis = 0,inplace=True) 
  return csv.values

def TimeArray(tags,time): #Taken directly from Wearable Dataset jupyter notebook
  array=np.insert(tags,0,time)
  Time_Array=[]
  for utc in array:
    tim=(datetime.datetime.strptime(utc,'%Y-%m-%d %H:%M:%S')-datetime.datetime.strptime(array[0], '%Y-%m-%d %H:%M:%S')).total_seconds()
    Time_Array.append(int(tim))
  return Time_Array
```

```python
blocks=['Baseline','REST1','REST2']
volunteers=os.listdir(f'{new_data}/REST')
if '.DS_Store' in volunteers: volunteers.remove('.DS_Store')
for volunteer in volunteers:
  for block in blocks:
    if not os.path.exists(f'{new_data}/REST/{volunteer}/{block}'):
      os.makedirs(f'{new_data}/REST/{volunteer}/{block}')
```

```python
#Get First and Second rest in stress protocol
protocol='REST'
volunteers=os.listdir(f'{new_data}/{protocol}')
if '.DS_Store' in volunteers: volunteers.remove('.DS_Store')
for volunteer in volunteers:
  for file in datafiles:
    data = pd.read_csv(f'Wearable_Dataset/STRESS/{volunteer}/{file}.csv')
    time = data.columns.values[0]
    data = removeSampRate(data)

    tags = pd.read_csv(f'Wearable_Dataset/STRESS/{volunteer}/tags.csv',header=None)
    tags = tags.values

    Time_Array = TimeArray(tags,time)

    mult = len(data)/Time_Array[-1]

    if 'S' in volunteer:
      Baseline = data[0:(round(Time_Array[2]*mult))]
    else:
      Baseline = data[0:(round(Time_Array[1]*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/Baseline/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(Baseline)

    if 'S' in volunteer:
      REST1 = data[(round(Time_Array[4]*mult)):(round(Time_Array[5]*mult))]
    else:
      REST1 = data[(round(Time_Array[3]*mult)):(round(Time_Array[4]*mult))]
    REST1 = REST1[len(REST1)//2:] #get second half
    with open(f'{new_data}/{protocol}/{volunteer}/REST1/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(REST1)

    if 'S' in volunteer:
      REST2 = data[(round(Time_Array[6]*mult)):(round(Time_Array[7]*mult))]
    else:
      REST2 = data[(round(Time_Array[7]*mult)):(round(Time_Array[8]*mult))]
    REST2 = REST2[len(REST2)//2:] #get second half
    with open(f'{new_data}/{protocol}/{volunteer}/REST2/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(REST2)
```

```python
blocks=['Stroop','TMCT','Real','Opposite','Subtract']
prot='STRESS'
volunteers=os.listdir(f'{new_data}/{prot}')
if '.DS_Store' in volunteers: volunteers.remove('.DS_Store')
for volunteer in volunteers:
  for block in blocks:
    if block=='Stroop' and 'f' in volunteer:
      continue
    if not os.path.exists(f'{new_data}/{prot}/{volunteer}/{block}'):
      os.makedirs(f'{new_data}/{prot}/{volunteer}/{block}')
```

```python
#Get stress sessions in stress protocol
protocol='STRESS'
volunteers=os.listdir(f'{new_data}/{protocol}')
if '.DS_Store' in volunteers: volunteers.remove('.DS_Store')
for volunteer in volunteers:
  for file in datafiles:
    data = pd.read_csv(f'Wearable_Dataset/{protocol}/{volunteer}/{file}.csv')
    time = data.columns.values[0]
    data = removeSampRate(data)

    tags = pd.read_csv(f'Wearable_Dataset/{protocol}/{volunteer}/tags.csv',header=None)
    tags = tags.values

    Time_Array = TimeArray(tags,time)

    mult = len(data)/Time_Array[-1]

    if 'S' in volunteer:
      Stroop = data[(round(Time_Array[3]*mult)):(round(Time_Array[4]*mult))]
      with open(f'{new_data}/{protocol}/{volunteer}/Stroop/{file}.csv', 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerows(Stroop)

    if 'S' in volunteer:
      TMCT = data[(round(Time_Array[5]*mult)):(round(Time_Array[6]*mult))]
    else:
      TMCT = data[(round(Time_Array[2]*mult)):(round(Time_Array[3]*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/TMCT/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(TMCT)

    if 'S' in volunteer:
      Real = data[(round(Time_Array[7]*mult)):(round(Time_Array[8]*mult))]
    else:
      Real = data[(round(Time_Array[4]*mult)):(round(Time_Array[5]*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/Real/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(Real)
    
    if 'S' in volunteer:
      Opposite = data[(round(Time_Array[9]*mult)):(round(Time_Array[10]*mult))]
    else:
      Opposite = data[(round(Time_Array[6]*mult)):(round(Time_Array[7]*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/Opposite/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(Opposite)
    
    if 'S' in volunteer:
      Subtract = data[(round(Time_Array[11]*mult)):(round(Time_Array[12]*mult))]
    else:
      Subtract = data[(round(Time_Array[8]*mult)):(round(Time_Array[9]*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/Subtract/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(Subtract)
```

```python
#Get aerobic sessions in aerobic protocol
protocol='AEROBIC'
volunteers=os.listdir(f'{new_data}/{protocol}')
if '.DS_Store' in volunteers: volunteers.remove('.DS_Store')
for volunteer in volunteers:
  for file in datafiles:
    data = pd.read_csv(f'Wearable_Dataset/{protocol}/{volunteer}/{file}.csv')
    time = data.columns.values[0]
    data = removeSampRate(data)

    tags = pd.read_csv(f'Wearable_Dataset/{protocol}/{volunteer}/tags.csv',header=None)
    tags = tags.values

    Time_Array = TimeArray(tags,time)

    mult = len(data)/Time_Array[-1]
    if file=='TEMP' and volunteer=='f08':
      print(Time_Array)

    if 'S' in volunteer:
      extract=((Time_Array[3]-Time_Array[2]-60)/2)
      rpm70 = data[(round((Time_Array[2]+extract)*mult)):(round((Time_Array[3]-extract)*mult))]
    else:
      extract=((Time_Array[2]-Time_Array[1]-60)/2)
      rpm70 = data[(round((Time_Array[1]+extract)*mult)):(round((Time_Array[2]-extract)*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/rpm70/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(rpm70)

    if 'S' in volunteer:
      extract=((Time_Array[4]-Time_Array[3]-60)/2)
      rpm75 = data[(round((Time_Array[3]+extract)*mult)):(round((Time_Array[4]-extract)*mult))]
    else:
      extract=((Time_Array[3]-Time_Array[2]-60)/2)
      rpm75 = data[(round((Time_Array[2]+extract)*mult)):(round((Time_Array[3]-extract)*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/rpm75/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(rpm75)

    if 'S' in volunteer:
      extract=((Time_Array[5]-Time_Array[4]-60)/2)
      rpm80 = data[(round((Time_Array[4]+extract)*mult)):(round((Time_Array[5]-extract)*mult))]
    else:
      extract=((Time_Array[4]-Time_Array[3]-60)/2)
      rpm80 = data[(round((Time_Array[3]+extract)*mult)):(round((Time_Array[4]-extract)*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/rpm80/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(rpm80)
    
    if 'S' in volunteer:
      extract=((Time_Array[6]-Time_Array[5]-60)/2)
      rpm85 = data[(round((Time_Array[5]+extract)*mult)):(round((Time_Array[6]-extract)*mult))]
    else:
      extract=((Time_Array[2]-Time_Array[1]-60)/2)
      rpm85 = data[(round((Time_Array[4]+extract)*mult)):(round((Time_Array[5]-extract)*mult))]
    with open(f'{new_data}/{protocol}/{volunteer}/rpm85/{file}.csv', 'w', newline='') as csvfile:
      writer = csv.writer(csvfile)
      writer.writerows(rpm85)
```
